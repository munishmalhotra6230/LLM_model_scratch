```python
# using dataset and data loader 
from torch.utils.data import Dataset,DataLoader
class GptDatasetv1(Dataset):
    def __init__(self,data,tokenizer,max_lenght,stride):
        super().__init__()
        self.input=[]
        self.target=[]
        token_ids=tokenizer.encode(data,allowed_special={"<|endoftext|>"})
        for i in range(0,len(token_ids)-max_lenght,stride):
            input_chunk=token_ids[i:i+max_lenght]
            target_chunk=token_ids[i+1:i+max_lenght+1]
            self.input.append(torch.tensor(input_chunk))
            self.target.append(torch.tensor(target_chunk))
    def __len__(self):
        return len(self.input)
    def __getitem__(self, index):
        return self.input[index],self.target[index]    
```

```python
import tiktoken
tokenizer=tiktoken.get_encoding('gpt2')
train_set=GptDatasetv1(raw_text,tokenizer,max_lenght=4,stride=4)
train_set_loader=DataLoader(train_set,8,True,drop_last=True)
data_iter=iter(train_set_loader)
inputs,target=next(data_iter)
```

```python
import torch
import torch.nn as nn

# encodings = tokenizer.encode("Hello this is Munish malhotra")
# encodingsx = torch.tensor(encodings).unsqueeze(0)

# # CORRECT FOR TIKTOKEN:
vocab_size = tokenizer.n_vocab 
token_embedding_layer = nn.Embedding(num_embeddings=vocab_size, embedding_dim=300)

# output = token_embedding_layer(encodingsx)
# output
# generating the tokens 
token_emb=token_embedding_layer(inputs)
token_emb.shape
```

```python
# Postional Encoding 
'''
we need embeddings in order to find the semantic meaning of the data 
see from above example "Hello this is Munish malhotra" now they converted into tokens like this 
[15496, 428, 318, 12107, 680, 6428, 8940, 430]
so in order to find the semantic meaning which means how these words like hello Munish like thing are corelated to each other 
but if postion dismatntel like munish is hello this doesnt show any meaning so we need the postions also 
'''
# there is two types of postional embeddings 
# absolute : adding aditional postional encodings to maintain the postional semantic meaning of vector 
# relative : this is interms of how far apart rather than exact position 

context_length=4
post_embed=torch.nn.Embedding(context_length,300)
postional_embedings=post_embed(torch.arange(context_length))
postional_embedings.shape
```

```python
input_embeddings=token_emb+postional_embedings
input_embeddings.shape
```
