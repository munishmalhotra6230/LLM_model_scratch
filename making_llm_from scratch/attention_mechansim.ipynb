```python
def softmax_naive(x):
    return torch.exp(x)/torch.exp(x).sum(dim=0)
print((softmax_naive(attn_scores)))# this is for journey 
```

```python
def attention_matrix(inputs):
    attention_weights=[]
    for i in range(len(inputs)):
        querx=inputs[i]
        attn_scores=torch.empty(inputs.shape[0])
        for j , x_i in enumerate(inputs):
            attn_scores[j]=torch.dot(x_i,querx)
        _scores= softmax_naive(attn_scores/sum(attn_scores))
        attention_weights.append(_scores) 
    return attention_weights   
```

```python
# now we will calculate the context vector 
import torch.nn as nn
class Self_attention(nn.Module):
    def __init__(self, input,dout):
        super().__init__()
        self.input =input
        # self.weights_query=nn.Parameter(torch.rand(input.shape[1],dout),requires_grad=True)
        # self.weights_key=nn.Parameter(torch.rand(input.shape[1],dout),requires_grad=True)
        # self.weights_value=nn.Parameter(torch.rand(input.shape[1],dout),requires_grad=True)
        self.weights_query=nn.Linear(input.shape[1],dout,bias=False)
        self.weights_key=nn.Linear(input.shape[1],dout,bias=False)
        self.weights_value=nn.Linear(input.shape[1],dout,bias=False)
        self.query=self.input@self.weights_query.weight.T
        self.keys=self.input@self.weights_key.weight.T
        self.value=self.input@self.weights_value.weight.T
        self.d_k=self.weights_key.weight.T.shape[1]
    def forward(self):
        self.attn_scores=self.query@self.keys.T
        self._attn_weights= torch.softmax(self.attn_scores/self.d_k**0.5,dim=-1)
        self.context_vector = self._attn_weights @ self.value
        return self.context_vector
 # here  in this version of code there is a catch for intializing the weights right now we are using the random numbers but it can further do more effiecently via intilising the linear        
```

# causal attention 

```python
batches=torch.stack((input_embeding,input_embeding))
```

```python
# this is called the masked attention 
class Causal_attention(nn.Module):
    def __init__(self, numemb,dout,drop_out=0.2,context_length=int):
        super().__init__()
        self.input =numemb
        self.dropout_rate=drop_out
        self.context_length=context_length
        # self.weights_query=nn.Parameter(torch.rand(input.shape[1],dout),requires_grad=True)
        # self.weights_key=nn.Parameter(torch.rand(input.shape[1],dout),requires_grad=True)
        # self.weights_value=nn.Parameter(torch.rand(input.shape[1],dout),requires_grad=True)
        self.weights_query=nn.Linear(self.input,dout,bias=False)
        self.weights_key=nn.Linear(self.input,dout,bias=False)
        self.weights_value=nn.Linear(self.input,dout,bias=False)
        self.d_k=self.weights_key.weight.T.shape[1]
        self.dropout=None
    def forward(self,inputs):# we can handle the data in batches 
        self.query=self.weights_query(inputs)
        self.keys=self.weights_key(inputs)
        self.value=self.weights_value(inputs)
        try:
           self.attn_scores=self.query@self.keys.transpose(1,2)
        except Exception as e :
            self.attn_scores=self.query@self.keys.T
        self.masked=torch.triu(self.attn_scores,diagonal=1)
        self.dropout=nn.Dropout(self.dropout_rate)
        self.casual_attn_scores=self.attn_scores.masked_fill(self.masked.bool()[:self.context_length,:self.context_length],-torch.inf)
        self.cas_attn_weights= torch.softmax(self.casual_attn_scores/self.d_k**0.5,dim=-1)# here 
        # normalising again 
        self.normalised_causal_attn=self.dropout(self.cas_attn_weights)
        return self.normalised_causal_attn @ self.keys
    # in this approach there is problem problem is that we are sharing the weights infulence via softmaxing the attn scores 
    # so this creates a data leakage problem in it : where as causal attention strictly dont allow to access the fututre tokens 
        
        
       
    
```

# multi head attention 

```python
# the idea is that the we can run the causal attention in parallel so that it can find complex pattern for words 

class Multi_Head_Wrapper(nn.Module):
    def __init__(self,din,dout,drop_out=0.2,num_heads=3,context_length=int):
        super().__init__()
        self.din=din
        self.dout=dout
        self.drop_=drop_out
        self.num_h=num_heads
        self.context_length=context_length
    def forward(self,x):
        self.head_cv=[]
        for i in range(self.num_h):
            self.context_vector=Causal_attention(self.din,self.dout,self.drop_,self.context_length).forward(x)
            self.head_cv.append(self.context_vector)
        return torch.cat(self.head_cv,dim=-1)
```

```python
class Multi_Head_Attention(nn.Module):
    def __init__(self,din_,dout,dropout_r,numh,bias=False):
        super().__init__()
        self.dout=dout
        self.heads=numh
        self.head_dim=self.dout//self.heads
        self.w_query=nn.Linear(din_,dout,bias=bias)
        self.w_key=nn.Linear(din_,dout,bias=bias)
        self.w_value=nn.Linear(din_,dout,bias=bias)
        self.out_proj=nn.Linear(dout,dout)
        self.dropout=nn.Dropout(dropout_r)
    def forward(self,x):
        batch,num_tokens,din_emb=x.shape
        query=self.w_query(x)
        key= self.w_key(x)
        value=self.w_value(x)
        key=key.view(batch,num_tokens,self.heads,self.head_dim)
        query=query.view(batch,num_tokens,self.heads,self.head_dim)
        value=value.view(batch,num_tokens,self.heads,self.head_dim)
        key=key.transpose(1,2)
        query=query.transpose(1,2)
        value=value.transpose(1,2)
        self.attn_scores=query @ key.transpose(2,3)
        masked=torch.triu(self.attn_scores,diagonal=1)
        attn_scoresx=self.attn_scores.masked_fill(masked.bool()[:num_tokens,:num_tokens],-torch.inf)
        attn_weight=torch.softmax(attn_scoresx/key.shape[-1],dim=-1)
        self.final_weights=self.dropout(attn_weight)
        self.context_vec=(self.final_weights @ value).transpose(1,2)
        self.context_vec_f=self.context_vec.contiguous().view(batch,num_tokens,self.dout)
        self.context_vec_f=self.out_proj(self.context_vec_f)
        return self.context_vec_f
    
```
